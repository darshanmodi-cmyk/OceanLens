```python
# --- 6.1 Clean Argo: drop bad QC and missing critical fields ---

# Argo QC convention: '1' = good data, '2' = probably good.
# Adjust the accepted set below if your files use a different QC scheme —
# check argo_eda["TEMP_QC"].unique() first if unsure.
GOOD_QC = {"1", "2"}

def qc_ok(series):
    return series.astype(str).str.strip().isin(GOOD_QC)

argo_clean = argo_eda.copy()

print("Before QC filter:", argo_clean.shape)
print("TEMP_QC values found:", argo_clean["TEMP_QC"].unique())

argo_clean = argo_clean[
    qc_ok(argo_clean["TEMP_QC"]) &
    qc_ok(argo_clean["PRES_QC"])
]

# Drop rows missing anything essential for training
argo_clean = argo_clean.dropna(
    subset=[
        "DATE",
        "LATITUDE (degree_north)",
        "LONGITUDE (degree_east)",
        "PRES (decibar)",
        "TEMP (degree_Celsius)"
    ]
)

print("After QC + null filter:", argo_clean.shape)
```

```python
# --- 6.2 Match each Argo row to its nearest OISST date ---

# `datasets` keys are the raw filename-derived date strings (e.g. "20260115").
# Parse them into real timestamps so we can compare against Argo dates.
# Parse the numeric date prefix and ignore suffixes such as "_preliminary"
oisst_date_lookup = {
    pd.to_datetime(str(k).split("_")[0], format="%Y%m%d"): k
    for k in datasets.keys()
}

oisst_date_index = sorted(oisst_date_lookup.keys())

def nearest_oisst_key(argo_date):
    argo_date = pd.Timestamp(argo_date)

    if argo_date.tzinfo is not None:
        argo_date = argo_date.tz_localize(None)

    nearest = min(
        oisst_date_index,
        key=lambda d: abs(argo_date - d)
    )

    return oisst_date_lookup[nearest]

argo_clean["OISST_DATE_KEY"] = argo_clean["DATE"].apply(nearest_oisst_key)

print(argo_clean["OISST_DATE_KEY"].value_counts())
oisst_date_index = sorted(oisst_date_lookup.keys())

def nearest_oisst_key(argo_date):
    argo_date = argo_date.tz_localize(None) if argo_date.tzinfo else argo_date
    diffs = [abs((argo_date - d)) for d in oisst_date_index]
    nearest = oisst_date_index[diffs.index(min(diffs))]
    return oisst_date_lookup[nearest]

argo_clean["OISST_DATE_KEY"] = argo_clean["DATE"].apply(nearest_oisst_key)

print(argo_clean["OISST_DATE_KEY"].value_counts())
```

```python
# --- 6.3 Vectorized nearest-neighbor SST lookup per matched date group ---

matched_frames = []

# SST matching is performed below, after argo_clean is rebuilt
# and OISST_DATE_KEY is created.

argo_clean = argo_eda.copy()

for column in ["TEMP_QC", "PRES_QC"]:
    argo_clean[column] = pd.to_numeric(argo_clean[column], errors="coerce")

argo_clean = argo_clean[
    argo_clean["TEMP_QC"].isin([1, 2]) &
    argo_clean["PRES_QC"].isin([1, 2])
].dropna(
    subset=[
        "DATE",
        "LATITUDE (degree_north)",
        "LONGITUDE (degree_east)",
        "PRES (decibar)",
        "TEMP (degree_Celsius)"
    ]
)

print("After QC + null filter:", argo_clean.shape)

# Recreate the OISST date key after rebuilding argo_clean
argo_clean["OISST_DATE_KEY"] = argo_clean["DATE"].apply(
    nearest_oisst_key
)

matched_frames = []

for date_key, group in argo_clean.groupby("OISST_DATE_KEY"):
    grid = datasets[date_key]

    lat_idx = xr.DataArray(
        group["LATITUDE (degree_north)"].to_numpy(),
        dims="points"
    )
    lon_idx = xr.DataArray(
        group["LONGITUDE (degree_east)"].to_numpy(),
        dims="points"
    )

    sst_values = grid.sel(
        lat=lat_idx,
        lon=lon_idx,
        method="nearest"
    ).to_numpy()

    matched_group = group.copy()
    matched_group["SST_SURFACE (degree_Celsius)"] = sst_values
    matched_frames.append(matched_group)

integrated_df = pd.concat(
    matched_frames,
    ignore_index=True
)

print("Integrated shape before final null drop:", integrated_df.shape)
print("Nulls in matched SST:", integrated_df["SST_SURFACE (degree_Celsius)"].isnull().sum())
```

```python
# --- 6.4 Final null cleanup (e.g. Argo points that landed on a masked/land pixel) ---

integrated_df = integrated_df.dropna(subset=["SST_SURFACE (degree_Celsius)"])

print("Final integrated shape:", integrated_df.shape)
integrated_df.isnull().sum()
```
